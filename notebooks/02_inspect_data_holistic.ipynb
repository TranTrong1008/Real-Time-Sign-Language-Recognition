```python
import numpy as np

data_dir = "../data/holistic/"

try:
   
    X_train = np.load(f"{data_dir}X_train_aug.npy")
    y_train = np.load(f"{data_dir}y_train_aug.npy")
    
    X_val = np.load(f"{data_dir}X_val.npy")
    y_val = np.load(f"{data_dir}y_val.npy")
    X_test = np.load(f"{data_dir}X_test.npy")
    y_test = np.load(f"{data_dir}y_test.npy")

    print("--- KIỂM TRA SHAPE DỮ LIỆU ---")
    print(f"X_train shape: {X_train.shape}")
    print(f"y_train shape: {y_train.shape}")
    print(f"X_val shape: {X_val.shape}")
    print(f"y_val shape: {y_val.shape}")
    print(f"X_test shape: {X_test.shape}")
    print(f"y_test shape: {y_test.shape}")

    assert X_train.shape[0] == y_train.shape[0], "LỖI: X_train và y_train không khớp số lượng mẫu!"
    assert X_val.shape[0] == y_val.shape[0], "LỖI: X_val và y_val không khớp số lượng mẫu!"
    assert X_test.shape[0] == y_test.shape[0], "LỖI: X_test và y_test không khớp số lượng mẫu!"
    print("\n-> [OK] Số lượng mẫu giữa X và y khớp nhau ở cả 3 tập.")

    # Xác minh Input Shape và Num Classes
    actual_input_shape = (X_train.shape[1], X_train.shape[2])
    
    # Kiểm tra định dạng nhãn (Label Encoding vs One-Hot)
    if len(y_train.shape) == 1 or y_train.shape[1] == 1:
        actual_num_classes = len(np.unique(y_train))
        label_format = "Label Encoding"
    else:
        actual_num_classes = y_train.shape[1]
        label_format = "One-Hot Encoding"

    print(f"\n--- XÁC MINH CẤU HÌNH ---")
    print(f"Thực tế INPUT_SHAPE: {actual_input_shape}")
    print(f"Thực tế NUM_CLASSES: {actual_num_classes}")
    print(f"Định dạng nhãn: {label_format}")
    
    # CẢNH BÁO QUAN TRỌNG:
    if label_format == "Label Encoding":
        print("\nNhãn là mảng 1 chiều (Label Encoding) khớp với sparse_categorical_crossentropy!")

except FileNotFoundError as e:
    print(f"LỖI: Không tìm thấy file dữ liệu. Chi tiết lỗi: {e}")
```
