```python
import numpy as np

# Đường dẫn file do M2 cung cấp (thay đổi nếu cần)
X_path = 'D:\#AI\A_Real_Time_Sign_Language_Recognition\Real-Time-Sign-Language-Recognition\data\holistic\X_train_aug.npy'
y_path = 'D:\#AI\A_Real_Time_Sign_Language_Recognition\Real-Time-Sign-Language-Recognition\data\holistic\y_train_aug.npy'
```

```python
# Sử dụng mmap_mode='r' để chỉ đọc metadata, không load toàn bộ file lớn vào RAM
X_data = np.load(X_path, mmap_mode='r')
y_data = np.load(y_path, mmap_mode='r')

print(" THÔNG TIN DỮ LIỆU ")
print(f"Data shape (X): {X_data.shape}")
print(f"Label shape (y): {y_data.shape}")
```

```python
num_samples = X_data.shape[0]
num_frames = X_data.shape[1]
num_features = X_data.shape[2]

print(f"\n1. Số lượng samples: {num_samples}")
print(f"2. Số frames mỗi sequence (T): {num_frames}")
print(f"3. Số features mỗi frame (F): {num_features}")

# Xác định số lượng class
if len(y_data.shape) > 1 and y_data.shape[1] > 1:
    num_classes = y_data.shape[1]
    print(f"4. Số lượng class (NUM_CLASSES): {num_classes} (Định dạng: One-hot encoding)")
else:
    num_classes = len(np.unique(y_data))
    print(f"4. Số lượng class (NUM_CLASSES): {num_classes} (Định dạng: Label encoding)")
```
